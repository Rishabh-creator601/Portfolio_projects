```python
class_list =  model.names  #{0: 'mask', 1: 'no-mask'}
```

```python
def track_images(image_path, model, class_list):
    coordinates = []
    # Run prediction and convert results to DataFrame
    results = model.predict(image_path)[0].boxes.data.detach().cpu().numpy()
    df = pd.DataFrame(results).astype("float")
    
    # Iterate through each detection
    for _, row in df.iterrows():
        x1, y1, x2, y2, class_id, conf = map(int, row)
        class_name = class_list[class_id]  # Optional: use it if needed
        coordinates.append([x1, y1, x2, y2])
    
    return (df,coordinates,class_name,conf)
```

```python
df,coordinates,class_name,conf =  track_images("./media/one.jpg",model,class_list)
```
