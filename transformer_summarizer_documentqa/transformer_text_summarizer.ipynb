```python
from transformers import T5Tokenizer, T5ForConditionalGeneration

model_name = 't5-small'
tokenizer = T5Tokenizer.from_pretrained(model_name)
model = T5ForConditionalGeneration.from_pretrained(model_name)
```

```python
def summarize(text):
    inputs = tokenizer.encode("summarize: " + text, return_tensors="pt", max_length=1024, truncation=True)
    summary_ids = model.generate(inputs, max_length=150, min_length=200, length_penalty=2.0, num_beams=4, early_stopping=True)
    return tokenizer.decode(summary_ids[0], skip_special_tokens=True)
```

```python
text =  """
In the hour before the world remembers its own heartbeat, early mornings build a kind of invisible architecture. Light isn’t fully born yet; it drifts like an undecided traveller, brushing rooftops and windowpanes as if choosing where to land. The air carries a gentle crispness that feels handcrafted, almost ceremonial, as though nature reset its canvas overnight.

Birds begin their soft rehearsals, not the full orchestra, just warm-up notes that hint at the score ahead. Streets, usually frenetic creatures by noon, lie still and contemplative, like giants enjoying a moment of peace before returning to their daily thunder. Even the trees seem taller at this time, their silhouettes stretching in quiet confidence.

There’s a peculiar clarity available in this pocket of the day. Thoughts line up neatly instead of tumbling over one another, ambitions feel lighter, and problems shrink to manageable pebbles. In these minutes, the world asks nothing from you; it simply invites you to exist, unhurried and whole.

The quiet architecture of early mornings is not carved in stone or steel but in stillness, breath, and intention. A gentle reminder that every day begins with a doorway, and stepping through it shapes everything that follows. 🌤️✨"""


```

```python
print(len(text))

text2  =  summarize(text)
print(text2)
len(text2)
```
